# SQL Sector

## 1. Create Table:

```python
import sqlite3

conn = sqlite3.connect('example.db')
cursor = conn.cursor()

cursor.execute('''
    CREATE TABLE IF NOT EXISTS users (id INTEGER PRIMARY KEY, name TEXT, age INTEGER)
''')

conn.close()
```

## 2. Insert Data:

```python
conn = sqlite3.connect('example.db')
cursor = conn.cursor()

cursor.execute('''
    INSERT INTO users (name, age) VALUES ('Alice', 30)
''')

conn.commit()
conn.close()
```

## 3. Select Data

```python
conn = sqlite3.connect('example.db')
cursor = conn.cursor()

cursor.execute('SELECT * FROM users')
rows = cursor.fetchall()

for row in rows:
    print(row)

conn.close()
```

## 4. Update Data

```python
conn = sqlite3.connect('example.db')
cursor = conn.cursor()

cursor.execute('''UPDATE users SET age = ? WHERE name = ?''', (35, 'Alice'))

conn.commit()
conn.close()
```

## 5. Delete Data:

```python
conn = sqlite3.connect('example.db')
cursor = conn.cursor()

cursor.execute('''
    DELETE FROM users WHERE name = ?
''', ('Alice',))

conn.commit()
conn.close()
```

#  Regular Expressions

Regular Expressions (Regex) are a powerful way to search for specific patterns within text or validate data against a certain format. They are used for searching, replacing, and extracting information from text.

How do they work?
A regular expression consists of a pattern that defines what you are looking for. It can include:

Plain characters (e.g., Hello to search for the word "Hello").
Special symbols (e.g., \d to search for any digit, or \s to search for spaces).

```python
import re

text = "رقم الهاتف هو 12345 6789"
pattern = r"\d+"  # البحث عن أي أرقام متتالية
match = re.search(pattern, text)

if match:
    print("Number Found", match.group())
```

```python
text = "بريدي الإلكتروني هو example@mail.com"
pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b"
match = re.search(pattern, text)

if match:
    print("تم العثور على بريد إلكتروني:", match.group())
```

## Regex is most useful when working with unstructured text from:

* User input (form validation, passwords, emails, etc.)
* Text files (logs, documents, CSV, config files, etc.)
* Web pages (HTML, scraping data, extracting links, etc.)
* Databases (cleaning messy text fields, extracting data patterns)
* Code analysis (searching function names, refactoring variables)
* 💡 If the data is structured (like JSON or SQL tables), it's better to use direct queries or libraries instead of Regex.

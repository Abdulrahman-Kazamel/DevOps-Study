```python
class Car:
    def __init__(self, brand):
        self.brand = brand  # public variable

    def show_brand(self):      # public method
        print(self.brand)

car1 = Car("Subaru")
car2 = Car("Porche")
car1.brand = "Fiat"
print(car1.brand)     # ✅ You can access it
print(car2.brand)
car2.show_brand()     # ✅ You can call this
```

```python
class Car:
    def __init__(self):
        self.__engine_number = "ABC123"  # private variable

    def __start_engine(self):           # private method
        print("Engine started")

    @property
    def drive(self):
        # Use private things **inside** the class
        print(f"Driving car with engine: {self.__engine_number}")
        self.__start_engine()

car = Car()

# ❌ You cannot access private things directly:
# print(car.__engine_number)     # Error
# car.__start_engine()           # Error

# ✅ But you can use a method that uses them internally:
car.drive
```

```python
class Car:
    def __init__(self):
        self._mileage = 10000  # protected variable
        # print(self._mileage)

car = Car()
car._mileage = 9999
print(car._mileage)  # ✅ This works, but it's a hint: "don’t touch this"
```

```python
class Car:
    def __init__(self):
        self._mileage = 10000

    @property
    def km(self):
        return self._mileage  # controlled access
```

```python
car = Car()
print(car.km)   # ✅ No need to call like a function
```

```python
car.mileage = 5000   # ❌ Error! because no setter
```

```python
class Car:
    def __init__(self):
        self._mileage = 10000

    @property
    def mileage(self):
        return self._mileage

    @mileage.setter
    def mileage(self, value):
        if value < 0:
            print("Mileage can't be negative!")
        else:
            self._mileage = value
```

```python
car = Car()
car.mileage = 5000
print(car._mileage)
```

```python
car = Car()
car._mileage = 5000   # ✅ Works, but not recommended
```
